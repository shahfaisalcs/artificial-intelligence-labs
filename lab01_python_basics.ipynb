### Q2 — Student Result System

```python
m1, m2, m3 = 85, 78, 90
average = (m1 + m2 + m3) / 3
if average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"
print("Average:", average)
print("Grade:", grade)
```

### Q3 — Login with at Most Three Attempts

```python
correct_password = "python123"
attempts = ["wrong1", "wrong2", "python123"]
logged_in = False
for attempt_number, password in enumerate(attempts[:3], start=1):
    if password == correct_password:
        print("Login successful on attempt", attempt_number)
        logged_in = True
        break
    else:
        print("Incorrect password")
if not logged_in:
    print("Maximum attempts reached")
```

### Q6 — Count Positive, Negative and Zero

```python
numbers = [5, -2, 0, 8, -1, 0, 3, -7, 4, 0]
positive = negative = zero = 0
for number in numbers:
    if number > 0:
        positive += 1
    elif number < 0:
        negative += 1
    else:
        zero += 1
print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

### Q7 — Prime Number Check

```python
number = 29
is_prime = True
if number < 2:
    is_prime = False
else:
    divisor = 2
    while divisor < number:
        if number % divisor == 0:
            is_prime = False
            break
        divisor += 1
if is_prime:
    print("Prime")
else:
    print("Not prime")
```
